```python
import pandas as pd
import os
```

```python
#group lab files
lab_path = "htk_files/data"
lab_files = []
for archivo in os.listdir(lab_path):
    ruta_completa = os.path.join(lab_path, archivo)
    if os.path.isfile(ruta_completa) and archivo.endswith('.lab'):
        lab_files.append(ruta_completa)
```

```python
full_df = pd.read_excel("global_table_con_UID.xlsx", sheet_name=0)
test_dfs = []
for i in range(1,11):
    test_dfs.append(pd.read_excel("global_table_con_UID.xlsx", sheet_name=i))
```

```python
#write scp files
for i, df in enumerate(test_dfs):
    scp_data = []
    j = 0
    while j < len(df):
        if df['Class'].iloc[j] == '2ESO':
            Class = 'SegundoESO'
        elif df['Class'].iloc[j] == '2PRIMARIA':
            Class = 'SegundoPRIMARIA'
        else:
            Class = 'QuintoPRIMARIA'
        scp_data.append(f"data/{Class}_{df['BPM'].iloc[j]}_{int(df['Kid number'].iloc[j]):02d}.htk")
        j+=1
    scp_filename = f"htk_files/trainings/Train{(i+1):02}/test{(i+1):02}.scp"
    with open(scp_filename, 'w', encoding='utf-8') as f:
        for scp in scp_data:
            f.write(f"{scp}\n")
```

```python
#write mlf files
for i, df in enumerate(test_dfs):
    test_labs = []
    j = 0
    while j < len(df):
        if df['Class'].iloc[j] == '2ESO':
            Class = 'SegundoESO'
        elif df['Class'].iloc[j] == '2PRIMARIA':
            Class = 'SegundoPRIMARIA'
        else:
            Class = 'QuintoPRIMARIA'
        test_labs.append(f"{Class}_{df['BPM'].iloc[j]}_{int(df['Kid number'].iloc[j]):02d}.lab")
        j+=1
    
    test_mlf_path = f"htk_files/trainings/Train{(i+1):02}/test{(i+1):02}.mlf"
    with open(test_mlf_path, 'w', encoding = 'utf-8') as f:
        f.write("#!MLF!#\n")
        for file in test_labs:
            f.write(f"\"*/{file}\"\n")
            filepath = next(archivo for archivo in lab_files if file in archivo)
            with open(filepath, 'r', encoding = 'utf-8') as f2:
                f.write(f"{f2.readline()}\n.\n")
```
